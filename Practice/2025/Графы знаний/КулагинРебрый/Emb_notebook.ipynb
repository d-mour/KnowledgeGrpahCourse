```python
g = rdflib.Graph()
g.parse("/content/onto-v3-populated.rdf", format="xml")

triples_list = []
for s, p, o in g:
    s_clean = str(s).split('#')[-1]
    p_clean = str(p).split('#')[-1]
    o_clean = str(o).split('#')[-1]
    triples_list.append([s_clean, p_clean, o_clean])

triples_arr = np.array(triples_list)
print(f"Загружено {len(triples_arr)} триплетов.")

tf_data = TriplesFactory.from_labeled_triples(triples_arr)
```

```python
result = pipeline(
    model='TransE',
    dataset_kwargs=None,
    training=tf_data,
    testing=tf_data,
    model_kwargs=dict(embedding_dim=50),
    training_kwargs=dict(num_epochs=1000, batch_size=256),
    random_seed=42,
)

model = result.model
```

```python
# Классификация

entity_to_id = training.entity_to_id

leader_ids = []
civ_ids = []

for s, p, o in triples_arr:
    if p == 'type':
        if s in entity_to_id:
            eid = entity_to_id[s]
            if o == 'Leader':
                leader_ids.append(eid)
            elif o == 'Civilization':
                civ_ids.append(eid)

leader_ids = list(set(leader_ids))
civ_ids = list(set(civ_ids))

print(f"Найдено лидеров: {len(leader_ids)}")
print(f"Найдено цивилизаций: {len(civ_ids)}")

# Формируем X и y
# X - эмбеддинги, y - метки классов (0 для Leader, 1 для Civilization)
all_embeddings = model.entity_representations[0](indices=None).detach().cpu().numpy()

if np.iscomplexobj(all_embeddings):
    all_embeddings = np.hstack([all_embeddings.real, all_embeddings.imag])

X_leaders = all_embeddings[leader_ids]
X_civs = all_embeddings[civ_ids]

X = np.vstack([X_leaders, X_civs])
y = np.array([0] * len(leader_ids) + [1] * len(civ_ids))

if len(X) > 10:
    # train/test
    X_train_clf, X_test_clf, y_train_clf, y_test_clf = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

    clf = XGBClassifier(eval_metric='logloss')
    clf.fit(X_train_clf, y_train_clf)

    y_pred = clf.predict(X_test_clf)
    acc = accuracy_score(y_test_clf, y_pred)

    print(f"\nТочность классификации: {acc:.2f}")

    dummy = DummyClassifier(strategy="most_frequent")
    dummy.fit(X_train_clf, y_train_clf)
    dummy_acc = dummy.score(X_test_clf, y_test_clf)
    print(f"Точность базовой модели: {dummy_acc:.2f}")

    if acc > dummy_acc:
        print("Успех: Эмбеддинги помогают различать типы сущностей лучше, чем случайность.")
    else:
        print("Результат на уровне базовой модели (возможно, классы слишком несбалансированы или мало данных).")

else:
    print("Слишком мало данных для обучения классификатора.")
```

```python
from sklearn.metrics.pairwise import cosine_similarity

leaders = set()
civilizations = set()

for s, p, o in triples_arr:
    if p == 'type' and o == 'Leader':
        leaders.add(s)
    if p == 'type' and o == 'Civilization':
        civilizations.add(s)


embeddings = model.entity_representations[0](indices=None).detach().cpu().numpy()

id_to_entity = {v: k for k, v in training.entity_to_id.items()}

def get_similar_entities(entity_name, top_k=5):
    if entity_name not in training.entity_to_id:
        print(f"Сущность '{entity_name}' не найдена в словаре.")
        return

    target_id = training.entity_to_id[entity_name]
    target_vec = embeddings[target_id].reshape(1, -1)

    sims = cosine_similarity(target_vec, embeddings)[0]

    sorted_indices = sims.argsort()[::-1]

    print(f"\nЕсли вам нравится '{entity_name}', попробуйте:")
    found = 0
    for idx in sorted_indices:
        label = id_to_entity[idx]

        if label == entity_name: continue

        is_game_actor = False

        if label in leaders or label in civilizations:
            is_game_actor = True
        elif "Civ6" in label:
            is_game_actor = True
        elif label in leaders:
             is_game_actor = True
        elif not any(x in label for x in ["Affinity", "Ability", "Structure", "Victory", "Affection"]):
             is_game_actor = True

        if is_game_actor:
            print(f"{found+1}. {label} (Сходство: {sims[idx]:.4f})")
            found += 1
            if found >= top_k: break

# --- Тестируем ---
get_similar_entities("German_Civ6")
get_similar_entities("Tokugawa")
get_similar_entities("Russian_Civ6")
```
