```python
from rdflib import URIRef, BNode, Literal, Namespace, Graph
from rdflib.namespace import Namespace, NamespaceManager
from rdflib.plugins import sparql
from rdflib.namespace import RDF, RDFS
from rdflib.serializer import Serializer
import json
from typing import List, Dict
import requests
import re
import time
import psycopg2
```

```python
# create empty graph
graph = Graph()
graph.parse('car_ontology.owl', format='turtle')
print(graph.serialize(format='n3'))
graph.serialize(format='turtle', destination='cars_onto_turt.rdf')
```

## Init Classes

```python
prefix = 'http://www.semanticweb.org/sofia/ontologies/2021/3/cars'
Car = URIRef(prefix + '#Car')
CarModel = URIRef(prefix + '#Model')
Manufacturer = URIRef(prefix + '#Manufacturer')
CarType = URIRef(prefix + '#CarType')

# car types
BudjetCar = URIRef(prefix + '#BudjetCar')
FamilyCar = URIRef(prefix + '#FamilyCar')
LuxCar = URIRef(prefix + '#LuxCar')
RareCar = URIRef(prefix + '#RareCar')
SmallCar = URIRef(prefix + '#SmallCar')
```

### Object properties

```python
CreatedBy = URIRef(prefix + '#createdBy')
IsOfModel = URIRef(prefix + '#isOfModel')
IsOfType = URIRef(prefix + '#isOfType')
```

```python
#model
Year_to = URIRef(prefix + '#year_to')
Year_from = URIRef(prefix + '#year_from')
IsPopular = URIRef(prefix + '#isPopular')
Name = URIRef(prefix + '#name')
# car

Clearence = URIRef(prefix + '#clearence')
CarClass = URIRef(prefix + '#class')
Height = URIRef(prefix + '#height')
Length = URIRef(prefix + '#length')
DoorsCount = URIRef(prefix + '#doorsCount')
AvailabilityCount = URIRef(prefix + '#availabilityCount')
Photo = URIRef(prefix + '#photo')
Width = URIRef(prefix + '#width')
Year = URIRef(prefix + '#year')
Country = URIRef(prefix + '#country')
FuelConsumption = URIRef(prefix + '#fuelConsumption')
BodyType = URIRef(prefix + "#bodyType")
Cost = URIRef(prefix + "#cost")
Power = URIRef(prefix + "#power")

#safety
Safety = URIRef(prefix + '#safety')
AOP = URIRef(prefix + '#AOP')
VRU = URIRef(prefix + '#VRU')
COP = URIRef(prefix + '#COP')
SA = URIRef(prefix + '#SA')
```

## Add Data

```python
# Add manufacturers to graph

def format(mark):
    mark_form = '_'.join(mark.split(' ')).upper()
    mark_form = '_'.join(mark_form.split('-')).upper()
    mark_form = '_'.join(mark_form.split('/')).upper()
    return mark_form

def get_manufacturers_list():
    f = open("Man.txt", "r")
    text = f.read()
    mans = re.split(r"\n", text)
    f.close()
    return mans

def add_manufacturers(manList):
    man_name_to_man: Dict = {}
    for mark in manList:
        mark_form = format(mark)
        manRef = URIRef(prefix + "#" + mark_form)
        graph.add((manRef, RDF.type, Manufacturer))
        man_name_to_man[mark] = manRef
    return man_name_to_man

manList = get_manufacturers_list()
man_name_to_man = add_manufacturers(manList)
```

```python
## DB

def connect_to_db():
    return psycopg2.connect(
        dbname='postgres',
        user='postgres',
        password='mysecretpassword',
        host='0.0.0.0')
```

```python
# add models

def get_mark_name(id):
    db_connection = connect_to_db()
    cursor = db_connection.cursor()
    query = "SELECT id, autoru_key from manufacturer where id=" + str(id)
    cursor.execute(query)
    rows = cursor.fetchall()
    cursor.close()
    db_connection.close()
    if len(rows) == 1:
        return rows[0][1]
    return None

def get_mark_id(name):
    db_connection = connect_to_db()
    cursor = db_connection.cursor()
    query = "SELECT id, autoru_key from manufacturer where autoru_key='" + format(name) + "'"
    cursor.execute(query)
    rows = cursor.fetchall()
    cursor.close()
    db_connection.close()
    if len(rows) == 1:
        return rows[0][0]
    return None

def get_uploaded_marks():
    db_connection = connect_to_db()
    cursor = db_connection.cursor()
    query = "SELECT distinct mark from car"
    cursor.execute(query)
    rows = cursor.fetchall()
    cursor.close()
    db_connection.close()
    return rows
    
def get_uri_from_name(name):
    if name:
        return URIRef(prefix + "#" + format(name))
    return None
    
def add_model_to_graph(row):
    specs: Dict = {}
    specs['manufacturer_id'] = row[0]
    specs['name'] = row[1]
    specs['count'] = row[2]
    specs['year_to'] = row[3]
    specs['year_from'] = row[4]
    specs['popular'] = row[5]
    my_mark = get_mark_name(specs['manufacturer_id'])
    if my_mark:
        modelRef = URIRef(prefix + "#" + format(specs['name']))
        graph.add((modelRef, RDF.type, CarModel))
        graph.add((modelRef, CreatedBy, get_uri_from_name(my_mark))) 
        graph.add((modelRef, Name, Literal(specs['name'])))
        graph.add((modelRef, AvailabilityCount, Literal(specs['count'])))
        graph.add((modelRef, Year_to, Literal(specs['year_to'])))
        graph.add((modelRef, Year_from, Literal(specs['year_from'])))
        graph.add((modelRef, IsPopular, Literal(specs['popular'])))
    
    
def add_all_models():
    db_connection = connect_to_db()
    cursor = db_connection.cursor()
    marks = get_uploaded_marks()
    for mark in marks:
        print(mark)
        if mark[0]:
            mark_id = get_mark_id(mark[0])
            fields = "manufacturer_id, name, count, year_to, year_from, popular"
            cursor.execute("SELECT " + fields + " from models where manufacturer_id='" + str(mark_id) + "'")
            rows = cursor.fetchall()
            print(len(rows))
            for row in rows:
                add_model_to_graph(row)
    cursor.close()
    db_connection.close()

add_all_models()
# get_uploaded_marks()
```

### Get car characteristics

```python
# retrieving car specification by mark, model, body type and year

def get_characteristics(row):
    specs: Dict = {}
    specs['mark'] = row[0]
    specs['model'] = row[1]
    specs['year'] = row[2]
    specs['clearance'] = row[3]
    specs['carClass'] = row[4]
    specs['height'] = row[5]
    specs['length'] = row[6]
    specs['doorsCount'] = row[7]
    specs['width'] = row[8]
    specs['fuelConsumption'] = row[9]
    specs['bodyType'] = row[10]
    specs['power'] = row[11]
    specs['cost'] = row[12]
    specs['availabilityCount'] = row[13]
    specs['photo'] = row[14]
    
    return specs
    
```

```python
# add models to graph


countries_map: Dict = {}
    
def add_car_to_graph(row):
    specs = get_characteristics(row)
    if not specs['mark'] or not specs['model']:
        return
    carRef = URIRef(prefix + "#" + format(specs['model']) + "_" + specs['bodyType'] + "_" + str(specs['year']))
    graph.add((carRef, RDF.type, Car))
    graph.add((carRef, IsOfModel, get_uri_from_name(specs['model'])))
    graph.add((carRef, Year, Literal(specs['year'])))
    if specs['clearance'] and specs['clearance'] > 0:
        graph.add((carRef, Clearence, Literal(specs['clearance'])))
    if specs['carClass'] and specs['carClass'] != '-1':
        graph.add((carRef, CarClass, Literal(specs['carClass'])))
    graph.add((carRef, Height, Literal(specs['height'])))
    graph.add((carRef, Length, Literal(specs['length'])))
    graph.add((carRef, DoorsCount, Literal(specs['doorsCount'])))
    graph.add((carRef, Width, Literal(specs['width'])))
    if specs['fuelConsumption'] and specs['fuelConsumption'] > 0:
        graph.add((carRef, FuelConsumption, Literal(specs['fuelConsumption'])))
    graph.add((carRef, BodyType, Literal(specs['bodyType'])))
    graph.add((carRef, Power, Literal(specs['power'])))
    
    graph.add((carRef, Cost, Literal(specs['cost'])))
    graph.add((carRef, AvailabilityCount, Literal(specs['availabilityCount'])))
    graph.add((carRef, Photo, Literal(specs['photo'])))
    
    
def add_all_cars():
    db_connection = connect_to_db()
    cursor = db_connection.cursor()
    fields = "mark, model, year, clearance, carClass, height, length, doorsCount, width, fuelConsumption, bodyType, power, cost, availabilityCount, photo"
    cursor.execute("SELECT " + fields + " from car")
    rows = cursor.fetchall()
    print(len(rows))
    for row in rows:
        add_car_to_graph(row)


add_all_cars()
```
